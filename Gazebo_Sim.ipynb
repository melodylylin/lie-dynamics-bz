```python
import pandas as pd
from matplotlib import pyplot as plt
names = ["pos_0.csv", "pos_7.csv", "pos_6.csv"]
for i,name in enumerate(names):
    df = pd.read_csv(name)
    plt.figure(1)
    
    if i == 0:
        plt.plot(df.PE[:5000], df.PN[:5000], '.', alpha = 0.1, color = "Red", label="High Fidelity")
    else:
        plt.plot(df.PE[:5000], df.PN[:5000], '.', alpha = 0.1, color = "Red")

plt.figure(1)
plt.legend()
plt.ylabel('North (m)')
plt.xlabel('East (m)')
plt.axis('equal')
plt.grid()
```
